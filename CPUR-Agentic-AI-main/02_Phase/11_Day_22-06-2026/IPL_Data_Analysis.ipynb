```python
df['team_pair'] = df.apply(lambda x: '-'.join(sorted([x['team1'], x['team2']])), axis=1)

pair_counts = df['team_pair'].value_counts().reset_index()
pair_counts.columns = ['team_pair', 'matches_played']

print(pair_counts.head(10))
```

```python
df['team_pair'] = df.apply(lambda x: '-'.join(sorted([x['team1'], x['team2']])), axis=1)
season_pairs = df.groupby(['season','team_pair']).size().reset_index(name='matches_played')
top_rivalries = season_pairs.sort_values(['season','matches_played'], ascending=[True,False])
for season in sorted(df['season'].unique()):
    print(f"\nSeason {season}:")
    print(top_rivalries[top_rivalries['season']==season].head(3))
```
