```python
# First, strip whitespace from column names just in case
df.columns = df.columns.str.strip()

# Convert to boolean (True if Graduate, False otherwise)
df['education'] = df['education'].str.strip() == 'Graduate'
```

```python
# Convert Boolean (True/False) to Integer (1/0)
df['education'] = df['education'].astype(int)
```

```python
# 1. Strip the hidden leading spaces from the text values
df['self_employed'] = df['self_employed'].str.strip()

# 2. Map 'Yes' to 1 and 'No' to 0
df['self_employed'] = df['self_employed'].map({'Yes': 1, 'No': 0})
```

```python
# 1. Clean the column names to remove hidden spaces
df.columns = df.columns.str.strip()

# 2. Combine all asset columns into a new column called 'total_assets'
df['total_assets'] = (
    df['residential_assets_value'] + 
    df['commercial_assets_value'] + 
    df['luxury_assets_value'] + 
    df['bank_asset_value']
)
```

```python
col = ["education","self_employed","income_annum","loan_amount","loan_term","cibil_score","total_assets","loan_status"]
df_clean = df[col].copy()
df_clean
```

```python
X = df[["education","self_employed","income_annum","loan_amount","loan_term","cibil_score","total_assets"]]
y = df["loan_status"]

model = LogisticRegression()
model.fit(X,y)

education = int(input("Enter education (1 for Graduate, 0 for Not Graduate): "))
self_employed = int(input("Are you self-employed? (1 for Yes, 0 for No): "))
income_annum = float(input("Enter annual income (e.g., 5000000): "))
loan_amount = float(input("Enter loan amount requested: "))
loan_term = int(input("Enter loan term in months: "))
cibil_score = int(input("Enter CIBIL/Credit score (300-900): "))
total_assets = float(input("Enter total value of all assets combined: "))



prediction = model.predict([[education,self_employed,income_annum,loan_amount,loan_term,cibil_score,total_assets]])
print("\n Prediction: ", prediction[0])
probability = model.predict_proba([[education,self_employed,income_annum,loan_amount,loan_term,cibil_score,total_assets]])

print("Rejected Probability = ", round(probability[0][0] * 100,2), "%")
print("Accepted Probability = ", round(probability[0][1] * 100,2), "%")

y_pred = model.predict(X)
accuracy = accuracy_score(y, y_pred)
print("Accuracy: ", accuracy * 100, "%")
```
