```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
data = {
    "Fever" : [98,99,100,101,102,103,98,99,101,102],
    "Cough" : [0,0,1,1,1,1,0,1,1,1],
    "BreathingProblem" : [0,0,0,1,1,1,0,0,1,1],
    "COVID_Status" : [
        "Negative","Negative","Negative",
        "Positive","Positive","Positive",
        "Negative","Negative","Positive","Positive"
    ]
}

df=pd.DataFrame(data)
```

```python
df
```

```python
df["COVID_Status"]=df["COVID_Status"].map({"Negative":0, "Positive":1})
df.astype(int)
```

```python
X = df[["Fever","Cough","BreathingProblem"]]
y = df["COVID_Status"]

model = DecisionTreeClassifier()
model.fit(X,y)

Fever = float(input("Enter your temprature: "))
cough = float(input("Enter your cough status (1 for yes) (0 for no) : "))
breathprob = float(input("do you have breathing rpoblem (1 for yes) (0 for no): "))

prediction = model.predict([[Fever,cough,breathprob]])[0]
probability = model.predict_proba([[Fever,cough,breathprob]])


print("\nProbability: ","postive" if prediction == 1 else "negative")
print("negative Probability = ", round(probability[0][0] * 100,2), "%")
print("Positive Probability = ", round(probability[0][1] * 100,2), "%")

y_pred = model.predict(X)
accuracy = accuracy_score(y, y_pred)
print("Accuracy: ", accuracy * 100, "%")
```
