```python
plt.figure(figsize=(8,5))
df['price'].hist(bins=50, color='skyblue')
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
df.groupby('bedrooms')['price'].mean().plot(kind='bar', color='orange')
plt.title("Average Price by Number of Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
plt.scatter(df['sqft_living'], df['price'], alpha=0.3, color='green')
plt.title("Living Area vs Price")
plt.xlabel("Square Feet of Living Area")
plt.ylabel("Price")
plt.show()
```

```python
df.groupby('grade')['price'].mean().plot(kind='line', marker='o', color='purple')
plt.title("Average Price by House Grade")
plt.xlabel("Grade")
plt.ylabel("Average Price")
plt.show()
```

```python
df.groupby('yr_built')['price'].mean().plot(kind='line', figsize=(10,5))
plt.title("Average Price by Year Built")
plt.xlabel("Year Built")
plt.ylabel("Average Price")
plt.show()
```
