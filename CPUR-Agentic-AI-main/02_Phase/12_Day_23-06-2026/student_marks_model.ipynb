```python
x=df[["number_courses","time_study"]]
y=df["Marks"]

model=LinearRegression()
model.fit(x,y)

print("slope: ", model.coef_[0])
print("slope: ", model.intercept_)

courses=int(input("Enter no of courses taken: "))
time=float(input("Enter no of hours studied: "))
 
marks=model.predict([[courses,time]])

result="pass" if marks>=35 else "fail"
print(f"predicted marks : = {marks[0]:.2f}")
print(f"predicted result: {result}")

plt.scatter(y, model.predict(x),label="Actual data")
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linestyle='--', label='Perfect Fit Line')
plt.xlabel("number_courses")
plt.ylabel("time_study")
plt.title("simple linear regression")
plt.legend()
plt.show()
```
