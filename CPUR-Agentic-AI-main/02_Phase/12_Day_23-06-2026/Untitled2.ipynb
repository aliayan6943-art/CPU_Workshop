```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = {
    "Area":     [1000,1200,1500,1800,2000,2200,2500,2800],
    "Bedrooms": [2,2,3,3,4,4,5,5],
    "Age":      [10,8,6,5,4,3,2,2],
    "Price":    [3000000,4000000,5000000,6000000,6500000,7000000,7500000,8000000]
}
df = pd.DataFrame(data)
print(df)

X = df[["Area", "Bedrooms", "Age"]]
y = df[["Price"]]

model = LinearRegression()
model.fit(X, y)

print("Slope: ", model.coef_)
print("Intercept: ", model.intercept_)

area = float(input("Enter Area (sq ft): "))
bedrooms = int(input("Enter Number of Bedrooms: "))
age = float(input("Enter Age of House (years): "))

price = model.predict(pd.DataFrame([[area, bedrooms, age]], columns=["Area","Bedrooms","Age"]))

print(f"Predicted Price for {area} sq ft, {bedrooms} bedrooms, {age} years old house = Rs.{price.item():.2f}")

plt.scatter(df["Area"], df["Price"], label="Actual Data")
plt.plot(df["Area"], model.predict(df[["Area","Bedrooms","Age"]]), label="Regression Line", color="red")
plt.xlabel("Area (sq ft)")
plt.ylabel("Price (Rs.)")
```
